### Exercise 1. Setup and load the data

Just like in the main notebook:
1. Install and import `BERTopic`.
2. Load the dataset into a dataframe called `df` (also import `pandas` if you need it).
3. Filter for rows where `abstract` is not null and `language` is `'en'`, and store the abstract texts in a list called `documents`.
4. Print how many documents you ended up with.

```python
# 1. install + import BERTopic
! pip install bertopic
from bertopic import BERTopic
import pandas as pd
# 2. load the dataset into df
df = pd.read_csv('/content/drive/MyDrive/dse-ml-2026/materials/datasets/dh_abstracts/dh2005-2026.csv')

# 3. filter for English, non-null abstracts -> documents
documents = df[(df.abstract.notnull()==True) & (df.language == 'en')].abstract.to_list()

# 4. print the number of documents
len(documents)
```

### Exercise 2. Train a topic model

Create a `BERTopic` model for English documents and fit it on `documents`, storing the results in `topics` and `probs`, like in the main notebook.

```python
# create topic_model
topic_model = BERTopic(verbose = True, language = 'english')

# fit_transform on documents -> topics, probs
topics, probs = topic_model.fit_transform(documents)
```

### Exercise 3. Find and inspect the biggest topic

The main notebook calls `get_topic(7)`, using a hardcoded topic id. This time, don't hardcode it:
1. Use `get_topic_info()` to get the topics table.
2. Programmatically find the id of the topic with the **most** documents, **excluding** the outlier topic `-1`.
3. Call `get_topic()` on that id to see its top words.

(Hint: `get_topic_info()` returns a regular pandas DataFrame, so you can filter and sort it like any other.)

```python
# 1. get the topics table
topic_table = topic_model.get_topic_info()
topic_table

# 2. find the id of the largest non-outlier topic
largest_topic_id = topic_table.loc[topic_table.loc[topic_table["Topic"] != -1, "Count"].idxmax(), "Topic"]
# 3. inspect its top words
topic_model.get_topic(largest_topic_id)
```
