### Pre-processing: punctuation removal
removes punctuation from a text

```python
# type a text containing punctuation
text = "Hello, world! How are you?"

# process the text with the SpaCy's NLP model
doc = nlp(text)

# keep only all tokens which are not a punctuation
tokens = [token.text for token in doc if not token.is_punct]

# print out the result
print(tokens)
```

Note: In a pipeline, however, you don't have to do this manually. During the vectorization, with the default settings, the punctuation is removed automatically.

### Pre-processing: stop-words removal
removes stop-words from a text (stop-words = frequent words that carry only a little meaning, such as articles or prepositions)

```python
# type a text containing stop-words
text = "This is a simple example of a sentence."

# process the text with an NLP model
doc = nlp(text)

# keep only tokens that are not stop-words
tokens = [token.text for token in doc if not token.is_stop]

# print out the result
print(tokens)
```

Note: In a pipeline, however, you don't have to do this manually. During the vectorization, you can choose to filter out the stop-words automatically.

### Pre-processing: lemmatization
turns the tokens into their base form, such as 'runs' --> 'run'

```python
# type a text with words using e.g., plurals, various tenses etc.
text = "The cats were running through the houses."

# process the text with an NLP model
doc = nlp(text)

# turn each token into its lemma
lemmas = [token.lemma_ for token in doc]

# print out the result
print(lemmas)
```

### Pre-processing: Stemming
SpaCy does not provide a built-in stemmer. If you would like to perform  stemming, you can use for example the Porter Stemmer form the nltk library.

```python
# import PorterStemmer from the nltk library
from nltk.stem import PorterStemmer

# initialize the stemmer
stemmer = PorterStemmer()

# type an example text
text = "The cats were running through the houses."

# process the text with the SpaCy's NLP model to get the tokens
doc = nlp(text)

# Turn the tokens that are not punctuation or spaces into their root forms
stems = [
    stemmer.stem(token.text)
    for token in doc
    if not token.is_punct and not token.is_space
]

# print out the results
print(stems)
```

### Pre-processing: POS-tagging
adds information about the part-of-speech of the token

```python
# type an example sentence
text = "The cats are running through the house."

#process the text with an NLP model
doc = nlp(text)

# print out each token together with its POS
for token in doc:
    print(token.text, token.pos_)
```

### Vectorization
reminder: Bag-of-Words representation of texts shows the frequencies for each token in each document in the form of a vector. 
We can adjust BoW by changing the parameters of CountVectorizer, a Python implementation of BoW.

```python
# import necessary libraries
from sklearn.feature_extraction.text import CountVectorizer
import pandas as pd

# type example texts
docs = [
    "A long time ago in a galaxy far, far away.",
    "The galaxy is far away."
]

# Initialize vectorizer. We are using the default parametres by leaving the brackets blank.
vectorizer = CountVectorizer()

# Fit and transform
X = vectorizer.fit_transform(docs)

# Convert to DataFrame for readability
df = pd.DataFrame(X.toarray(), columns=vectorizer.get_feature_names_out())
print(df)
```

```python
# to use binary BoW, we set the 'binary' parameter to 'true':

vectorizer = CountVectorizer(binary=True)
X = vectorizer.fit_transform(docs)
pd.DataFrame(X.toarray(), columns=vectorizer.get_feature_names_out())
```

```python
# To specify that we want to use n-grams, we need to specify the n-gram range:

vectorizer = CountVectorizer(ngram_range=(1,2))  # unigrams + bigrams
X = vectorizer.fit_transform(docs)
pd.DataFrame(X.toarray(), columns=vectorizer.get_feature_names_out())
```

```python
# To remove stopwords, we specify we want to use the built-in english list of stop words. 
# To use other language, you would need to provide a custom list of stop-words.

vectorizer = CountVectorizer(stop_words='english')
X = vectorizer.fit_transform(docs)
pd.DataFrame(X.toarray(), columns=vectorizer.get_feature_names_out())
```

```python
# here is an example of how to use a custom German stop-words list:

docs = [
    'Eine Katze hat gerade auf dem Polster geschlafen.',
    'Wir sind gerade an der Uni in Graz.'
]

german_stopwords = [
    "der", "die", "das", "den", "dem", "des",
    "ein", "eine", "einer", "einem", "einen",
    "und", "oder", "aber", "ist", "sind",
    "an", "auf", "in"
]

vectorizer = CountVectorizer(stop_words=german_stopwords)

X = vectorizer.fit_transform(docs)
pd.DataFrame(X.toarray(), columns=vectorizer.get_feature_names_out())
```

```python
# restricting vocabulary size - keeping only 5 most common tokens:
vectorizer = CountVectorizer(stop_words=german_stopwords, max_features=5)
X = vectorizer.fit_transform(docs)
pd.DataFrame(X.toarray(), columns=vectorizer.get_feature_names_out())
```

### BoW is not just a theory! 
It is a useful representation in plenty of tasks, including spam filtering:

```python
# Step 1: Import required libraries
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
```

```python
# Step 2: Define a small dataset (5 spam + 5 non-spam emails)
emails = [
    "Congratulations! You've won a free iPhone!",        # spam
    "Limited offer: Claim your $1000 gift card now!",    # spam
    "Win money now!!!",                                  # spam
    "Get rich quick with this simple trick",             # spam
    "Act now! Limited time deal on lottery tickets!",    # spam
    "Meeting scheduled at 10 AM tomorrow",               # not spam
    "Don't forget to send the report",                   # not spam
    "Lunch at 12?",                                      # not spam
    "Please review the attached document",               # not spam
    "Let’s discuss the project update",                  # not spam
]

labels = [1, 1, 1, 1, 1,   # 1 = spam
          0, 0, 0, 0, 0]   # 0 = not spam
```

```python
# Step 3: Split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    emails, labels, test_size=0.3, random_state=42, stratify=labels  
)
```

```python
# check out the lenghts of the vectors
print(len(X_train))
print(len(y_train))

print(len(X_test))
print(len(y_test))
```

```python
# Step 4: Convert text to feature vectors using CountVectorizer
vectorizer = CountVectorizer()
X_train_vec = vectorizer.fit_transform(X_train)
X_test_vec = vectorizer.transform(X_test)
```

```python
# Let's have a look inside!
pd.DataFrame(X_train_vec.toarray(), columns=vectorizer.get_feature_names_out())
```

```python
pd.DataFrame(X_test_vec.toarray(), columns=vectorizer.get_feature_names_out())
```

```python
# Step 5: Train an SVM classifier
model = SVC(kernel='linear')  # simple linear SVM
model.fit(X_train_vec, y_train)
```

```python
# Step 6: Predict on test data
y_pred = model.predict(X_test_vec)
```

```python
# Step 7: Evaluate the model
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)
```

```python
# Step 8: Print evaluation results
print("📊 Evaluation Metrics on Test Set:")
print(f"Accuracy:  {accuracy:.2f}")
print(f"Precision: {precision:.2f}")
print(f"Recall:    {recall:.2f}")
print(f"F1 Score:  {f1:.2f}")
```

```python
# Bonus: Try out with your own mail for prediction
my_input = ['Humanitarian Grant of 2M for you, contact me for quick claims.']
X_test_vec = vectorizer.transform(my_input)
y_pred = model.predict(X_test_vec)
```
